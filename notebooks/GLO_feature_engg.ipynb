```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("post_eda_glo.csv")
```

```python
df["EXPANSION_PERCENTILE"] = (
    df["EXPANSION_RATE"]
    .rank(pct=True) * 100
)
```

```python
df.head(10)
```

```python
df["AREA_PERCENTILE"] = (
    df["AREA_DISSOLVED"]
    .rank(pct=True) * 100
)
```

```python
df.head(10)
```

```python
df["GLACIER_CONNECTIVITY_SCORE"] = (
    df["CONNECTIVITY"]
    .eq("Glacier-fed")
    .astype(int) * 100
)
df.head()
```

```python
df["SIGNIFICANCE_SCORE"] = (
    df["EXPANSION_RATE_SIG"]
    .notna()
    .astype(int) * 100
)
```

```python
df.head()
```

```python
df1 = df.copy()
cluster_features = df[['LOG_AREA',
                    'EXPANSION_PERCENTILE',
                    'AREA_PERCENTILE',
                    'GLACIER_CONNECTIVITY_SCORE',
                    'SIGNIFICANCE_SCORE']]
```

```python
cluster_features.head()
```

```python
cluster_features.to_csv(
    "processed_ufc_fights\.csv",
    index=False
)
```
